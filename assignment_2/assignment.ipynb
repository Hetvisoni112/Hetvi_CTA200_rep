```python
#Part 1
def f(x):
    """
    This function returns the value (x ^ 3) - (x ^ 2) - 1  
    Parameters
    ----------- 
    x: float or integer
    
    Returns
    -----------
    float or integer
        float if (x ^ 3) - (x ^ 2) - 1 is not a whole number, othersise integer
    """
    return (x**3) - (x**2) - 1

def df(x):
    """
    This function returns the value 3 * (x ^ 2) - (2 * x)
    Parameters: x ----------- float or integer
    Returns: 
    3 * (x ^ 2) - (2 * x) ----------- float or integer
    """
    return 3*(x**2) - (2*x)
```

```python
# Part 2
def newton(f, df, x0, epsilon=1e-6, max_iter=30):
    """
    This function finds a root of f using Newton's method.

    Starting from an initial guess x0, iteratively applies the update
    rule x_(n+1) = x_n - f(x_n) / f'(x_n) until convergence or until
    the maximum number of iterations is reached.

    Parameters
    ----------
    f : callable
        The function whose root is to be found.
    df : callable
        The derivative of f.
    x0 : float
        Initial guess for the root.
    epsilon : float, optional
        Convergence tolerance. Iteration stops when |f(x_n)| < epsilon.
        Default is 1e-6.
    max_iter : int, optional
        Maximum number of iterations allowed. Default is 30.

    Returns
    -------
    float or None
        The approximated root x_n if |f(x_n)| < epsilon, otherwise None.
    """
    xn = x0
    if df(xn) == 0:
        print(f"Division by zero error: df({xn}) = 0")
        return None
    for n in range(1, max_iter + 1):
        xn = xn - f(xn) / df(xn)
        if abs(f(xn)) < epsilon:
            print(f"Found root in {n} iterations")
            return xn

    print("Iteration failed")
    return None
```
